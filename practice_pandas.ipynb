```python
# In This Part you will learn:
# Missing Values
# Detect Missing values
# Fill Missing Values
# Remove missing values
# Find Duplicates
# Remove duplcates

# Q1. What is missing values?
# A1. Missing values occur when some data is unavailable.
# Example:

import pandas as pd
data = {
    "Product": ["Laptop", "Mobile", "None", "Tablet"],
    "Price"  : [50000,None,18000,15000]
} 
df = pd.DataFrame(data)
print(df)
```

```python
# 2.Detect missing values
# Use isnull() to identify missing values
# Example 
# print(df.isnull)
# count missing values in each column:
# print(df.isnull().sum())
```

```python
# Import pandas
import pandas as pd

# Create sample data (with missing values)
data = {
    "Name": ["Ali", "Ahmed", None, "Usman"],
    "Age": [20, None, 22, 21],
    "Marks": [85, 90, None, 88]
}

# Convert into DataFrame
df = pd.DataFrame(data)

# Display DataFrame
print("Original DataFrame:\n")
print(df)

# 1. Detect missing values (True/False)
print("\nMissing Values (True means missing):\n")
print(df.isnull())

# 2. Count missing values in each column
print("\nCount of missing values in each column:\n")
print(df.isnull().sum())
```

```python
# import pandas
import pandas as pd

# create sample data (with missing values)

data = {
    "Product":["Laptop", None, "Mobile", "Tablet", None],
    "Price" :[None, 2500, 25000, None,30000],
    "Stock" : [25, 15, None, 56, 67]
}
df = pd.DataFrame(data)

print("\n Orignal data frame: \n")
print(df)

print("\n Missing Values( True mean missing):\n")
print(df.isnull())

print("\n Missing values in each column: \n")
print(df.isnull().sum())
```

```python
# 3. remove missing values
# Use dropna() to remove rows with missing values.
# Example :
# clean_df = df.dropna()
# print(clean_df) Only rows with complete data will remain
```

```python
# Import pandas
import pandas as pd

# Create sample data (with missing values)
data = {
    "Name": ["Ali", "Ahmed", None, "Usman"],
    "Age": [20, None, 22, 21],
    "Marks": [85, 90, None, 88]
}

# Convert into DataFrame
df = pd.DataFrame(data)

# Display original DataFrame
print("Original DataFrame:\n")
print(df)

# Remove rows with missing values
clean_df = df.dropna()

# Display cleaned DataFrame
print("\nDataFrame after removing missing values:\n")
print(clean_df)
```

```python
import pandas as pd

data = {
    "Name" : ["Hassan",None, "Muhammad", "Hassan", None],
    "Age" : [19, 45, None, 19, 78],
    "Marks": [56,78,89,56,None]
}
df = pd.DataFrame(data)
df["Name"] = df["Name"].fillna("Ahmad")
df["Name"] = df["Name"].fillna("Jawad")
df["Age"] = df["Age"].fillna(67)
df["Marks"] =df["Marks"].fillna(67)

print("\nOrignal dataframe: \n")
print(df)

print("\n Missing Values: (Trues mean missing values): \n")
print(df.isnull())

print("\n Missing values in each column: \n")
print(df.isnull().sum())

print("\n Dataframe after removing missing values: \n")
print(df.dropna())

print("\n duplicates values in all Rows:\n")
print(df.duplicated())

print("\n duplocates values in columns: \n")
print(df.duplicated(subset="Name"))

print("\n remove duplicates: \n")
print(df.drop_duplicates())

print("\nCheck Data information to understand dataset: \n")
print(df.info()) 

print("\n Statistical information (count, mean, standard deviation, minimum ,maximun, quartiles )")
print(df.describe())
```
